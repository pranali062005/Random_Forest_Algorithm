```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report
from sklearn.preprocessing import LabelEncoder
```

```python
import kagglehub

# Download latest version
path = kagglehub.dataset_download("amaymishra11/tech-layoffs-and-hiring-trends-2026")

print("Path to dataset files:", path)
```

```python
df = pd.read_csv(f'{path}/tech_layoffs_hiring_trends_elite_v2.csv')
```

```python
df = df.drop(columns='record_id')
```

```python
OBJ = LabelEncoder()

df['company_name'] = OBJ.fit_transform(df['company_name'])
df['industry'] = OBJ.fit_transform(df['industry'])
df['country'] = OBJ.fit_transform(df['country'])
df['month'] = OBJ.fit_transform(df['month'])
df['year'] = OBJ.fit_transform(df['year'])
df['layoffs_count'] = OBJ.fit_transform(df['layoffs_count'])
df['layoff_percentage'] = OBJ.fit_transform(df['layoff_percentage'])
df['reason_for_layoffs'] = OBJ.fit_transform(df['reason_for_layoffs'])
df['ai_automation_impact'] = OBJ.fit_transform(df['ai_automation_impact'])
df['ai_replacement_risk'] = OBJ.fit_transform(df['ai_replacement_risk'])
df['open_roles'] = OBJ.fit_transform(df['open_roles'])
df['hiring_trend'] = OBJ.fit_transform(df['hiring_trend'])
df['remote_jobs_percentage'] = OBJ.fit_transform(df['remote_jobs_percentage'])
df['top_hiring_role'] = OBJ.fit_transform(df['top_hiring_role'])
df['stock_growth_percent'] = OBJ.fit_transform(df['stock_growth_percent'])
df['revenue_growth_percent'] = OBJ.fit_transform(df['revenue_growth_percent'])
df['salary_budget_change'] = OBJ.fit_transform(df['salary_budget_change'])
df['ai_adoption_level'] = OBJ.fit_transform(df['ai_adoption_level'])
df['employee_sentiment'] = OBJ.fit_transform(df['employee_sentiment'])
df['job_security_score'] = OBJ.fit_transform(df['job_security_score'])
df['market_condition'] = OBJ.fit_transform(df['market_condition'])
df['company_size'] = OBJ.fit_transform(df['company_size'])
```

```python
df.columns
```

```python
x = df[['company_name', 'industry', 'country', 'company_size', 'month', 'year',
       'layoffs_count', 'layoff_percentage', 'reason_for_layoffs',
       'ai_automation_impact', 'ai_replacement_risk', 'open_roles',
       'hiring_trend', 'remote_jobs_percentage', 'top_hiring_role',
       'stock_growth_percent', 'revenue_growth_percent',
       'salary_budget_change', 'ai_adoption_level', 'employee_sentiment',
       'job_security_score']]

y = df['market_condition']
```

```python
x_train , x_test ,y_train , y_test = train_test_split(x,y,test_size = 0.2 , random_state = 42)
```

```python
model = RandomForestClassifier(
    n_estimators=50
)
model.fit(x_train,y_train)
```

```python
df.columns
```

```python
predict = model.predict(x_test)
predict
```

```python
from sklearn.metrics import accuracy_score

score = accuracy_score(y_test, predict)
print("Accuracy:", score)
```

```python
import pickle

pickle.dump(model, open('RandomForest.pkl', 'wb'))
print('PKL File Created Successfully')
```
